## Entramos a shell para revisar los datos que existen actualmente en el DB

Invocamos al modelo y mandamos a llamar todos los registros que existan

## Ocupamos el siguiente código para generar los 500 registros nuevos

```python
products_data = []
for i in range(1,501): 
    new_data = {"title":"Producto Prueba {}".format(i), "price":i*100+.99, "description":"Descripcion Prueba", "seller": "Vendedor Prueba", "color":"N/A", "product_dimensions":"N/A"}
    products_data.append(new_data)
```

## Insertamos y guardamos los productos creados en la base de datos

```python
from ecommerce.models import ProductModel
new_objects= []
for product_data in products_data:
    print(product_data)
    new_objects.append(ProductModel(**product_data))

ProductModel.objects.bulk_create(new_objects, ignore_conflicts=True)
```

## Revisamos el Admin para ver los productos que se han guardado exitosamente

Dentro de la carpeta ecommerce/fixtures/products creamos el archivo llamado "500Products.json"

Realizamos el fixture del archivo con el siguiente comando:

```python
python manage.py dumpdata ecommerce --indent 4 --format json > ecommerce/fixtures/products/500Products.json
```
